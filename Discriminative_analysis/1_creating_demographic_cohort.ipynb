```python
import pandas as pd
import numpy as np
```

```python
# reading in dataset

demo = pd.read_csv('ClinicalV1_anonymized.csv')
```

```python
len(demo)
```

```python
# Filtering for comorbidities (PD, FTD, ALS, TBI, stroke, TIA)

demo_filtered = demo[(demo['pd'] == 0) & (demo['ftd'] == 0) & (demo['als'] == 0) & (demo['tbi'] == 0) &
                     (demo['stroke'] == 0) & (demo['tia'] == 0)]
```

```python
len(demo_filtered)
```

```python
# Separating ad participants
ad_group = demo_filtered[demo_filtered['ad'] == 1]
```

```python
len(ad_group)
```

```python
potential_controls = demo_filtered[demo_filtered['ad'] == 0]
```

```python
# Assessing how many have the MMSE
```

```python
potential_controls[['cognitive_test_battery']].value_counts()

# Removing CAR scores- less well known cognitive battery
```

```python
controls_cognition = potential_controls[potential_controls['cognitive_test_battery'] != 'CAR']
```

```python
controls_cognition[['cognitive_test_battery']].value_counts()
```

```python
# Splitting those who have the MMSE from those who have the MOCA

MMSE = controls_cognition[controls_cognition['cognitive_test_battery'] == 'MMSE']
MOCA = controls_cognition[controls_cognition['cognitive_test_battery'] == 'MOCA']
```

```python
# Filtering MMSE

MMSE_cognitively_normal = MMSE[MMSE['cognitive_test_score'] >= 26]
```

```python
len(MMSE_cognitively_normal)
```

```python
MOCA_cognitively_normal = MOCA[MOCA['cognitive_test_score'] >= 17]
```

```python
MOCA_cognitively_normal
```

```python
MOCA['cognitive_test_score'].describe()

# MOCA scores unavailable for most participants -> will remove and just use MMSE
```

```python
# Further filtering with computed cognitive impairment

cohort = MMSE_cognitively_normal[MMSE_cognitively_normal['computed_cognitive_impairment'] == 0]
```

```python
len(cohort)

# 2514 controls
```

```python
cohort['age_at_visit'].describe()
```

```python
# Adding to full dataset

final_cohort = pd.concat([ad_group , cohort])
```

```python
len(final_cohort)
```

```python
final_cohort['ad'].value_counts()
```

```python
final_cohort.to_csv('initial_cohort.csv')
```
