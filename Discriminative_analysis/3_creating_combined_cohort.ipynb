```python
# importing libraries

import numpy as np
import pandas as pd
```

```python
# Reading in my two datasets I created

demo = pd.read_csv('initial_cohort.csv')
prot = pd.read_csv('protein_df.csv')
                
```

```python
# combining demographics and protein data
final_df = pd.merge(demo, prot, on='sample_id', how='inner')
```

```python
# Dropping all non-baseline visits

final_df_baseline = final_df[final_df['sequential_visit_number'] == 1]
```

```python
len(final_df_baseline)
```

```python
# Dropping unneeded columns

final_df_clean = final_df_baseline.drop(
    labels= ['Unnamed: 0' , 'contributor_code' , 'sequential_visit_number' , 'computed_age_range' , 
             'race' , 'years_of_education' , 'computed_years_education_range', 'computed_height_range', 
             'computed_weight_range', 'bmi', 'computed_bmi_range', 'resting_heart_rate_pulse', 'computed_pulse_range', 
             'systolic_blood_pressure_sitting', 'computed_systolic_blood_pressure_range', 
             'diastolic_blood_pressure_sitting', 'computed_diastolic_blood_pressure_range', 'alcohol_hx', 'smoking_hx',
             'total_years_smoked', 'computed_years_smoked', 'stroke', 'tia', 'tbi', 'recruited_control', 'ftd', 'pd', 
             'als', 'mci_sci', 'cancer', 'diabetes', 'chf', 'copd', 'mi', 'afib', 'angina', 'hyperlipidaemia',
             'hypertension', 'depression', 'depression_inventory_score', 'depression_test_battery', 'anxiety', 'cdr',
             'cognitive_test_date', 'cognitive_test_score', 'computed_cognitive_test_score', 'cognitive_test_battery',
             'computed_clinical_diagnosis', 'computed_cognitive_impairment', 'is_neuropath', 'is_biomarker'], 
    axis= 1)
```

```python
# Removing NA values (-1)

final_df_age = final_df_clean[final_df_clean['age_at_visit'] > 0]

final_df_age['age_at_visit'].describe()
```

```python
final_df_age['sex'].value_counts()
```

```python
# Removing NA values

final_df_sex = final_df_age[final_df_age['sex'] > 0]

final_df_sex['sex'].value_counts()
```

```python
# Separating ad and hc to generate some descriptives 

ad = final_df_sex[final_df_sex['ad'] == 1]
hc = final_df_sex[final_df_sex['ad'] == 0]
```
